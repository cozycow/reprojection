```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
import glob
from reprojection import *
from view import *
from show import *
from utils import *
from transforms import *
```

```python
files_fdt = sorted(glob.glob('/home/ulyanov/data/pmi/20250919/fdt/*'))
```

```python
with fits.open(files_fdt[-1]) as hdul:
    header_fdt = hdul[0].header
    data_fdt = hdul[0].data

header_fdt['CROTA'] += 0.25
```

```python
nx, ny = 1024, 1024
xc, yc = (nx - 1) / 2, (ny - 1) / 2
rsun = 0.45 * nx

data_fdt_, alpha_fdt, view_fdt = reproject(data_fdt, header_fdt, kind='',
                                           nx=nx, ny=ny, xc=xc, yc=yc, rsun=rsun,
                                           crota=0, crlt=0, rsun_arc=0,
                                           minmu=0.)
```

```python
show_data(data_fdt_, view_fdt,
          title='FDT', label=r'$B_{los}$, G',
          cmap='seismic', vmin=-200, vmax=200)
```

```python
files_hmi = sorted(glob.glob('/home/ulyanov/data/pmi/20250919/hmi/*'))
```

```python
with fits.open(files_hmi[0]) as hdul:
    header_hmi = hdul[1].header
    data_hmi = hdul[1].data

data_hmi = rebin(data_hmi, 4, update_header=header_hmi)
```

```python
show_data(data_hmi_, view_hmi,
          title='HMI', label=r'$B_{los}$, G',
          cmap='seismic', vmin=-200, vmax=200)
```
