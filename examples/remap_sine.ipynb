Mapping and showing the data in spherical coordinates

```python
data1 = remap(data, header, dlat=1 / 359.5, dlon=0.25, correct_mu=True, mu_thr=0.1, correct_dr=True, sine_lat=True)
```

```python
def show_map(data, figsize=(15,8), label=r'$B_{r}$, G', cmap='seismic', vmin=-50, vmax=50, **kwargs):
    fig, ax = plt.subplots(figsize=figsize)
    im = ax.imshow(data, origin='lower', cmap=cmap, vmin=vmin, vmax=vmax,
                   extent=[0, 360, -1, 1], aspect='auto', **kwargs)

    ax.set_xticks(np.arange(0,361,30), np.arange(0,361,30))
    ax.set_yticks(np.sin(np.arange(-90, 91, 15) * np.pi / 180), np.arange(-90, 91, 15))

    ax.set_xlabel('Longitude, degrees')
    ax.set_ylabel('Latitude, degrees')

    cax = ax.inset_axes((0.03, 0.1, 0.1, 0.015))
    fig.colorbar(im, cax=cax, orientation='horizontal', label=label)

    ax.grid(True, ls='--', color='black', alpha=0.2)
    plt.tight_layout()
```

```python
show_map(data1, vmin=-100, vmax=100)

nx, ny = data1.shape
nx, ny
```
