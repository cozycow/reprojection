```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
import glob
from reprojection import *
from show import *
from utils import *
```

```python
#files = sorted(glob.glob('/home/ulyanov/data/solo/phi/2024_2025/*blos_202505*'))
files = sorted(glob.glob('/home/ulyanov/data/solo/phi/2026/blos_/*'))
files
```

```python
output_file = 'wcs.csv'

with open(output_file, 'w') as f:
    f.write('date, did, dsun_au, car_rot, crln_obs, hgln_obs, crlt_obs, obs_vr, obs_vw, obs_vn, fgh_tsp, contposn\n')

for file in files[:]:
    with fits.open(file) as hdul:
        header = hdul[0].header.copy()

    file_ = file.split('/')[-1]

    date = file_.split('_')[-3]
    did = file_.split('_')[-1].split('.')[0]


    dsun_au = header['DSUN_AU'] if 'DSUN_AU' in header else header['DSUN_OBS'] / header['DSUN_REF']
    car_rot = header['CAR_ROT']
    crln_obs = header['CRLN_OBS']
    hgln_obs = header['HGLN_OBS'] if 'HGLN_OBS' in header else 0
    crlt_obs = header['CRLT_OBS']
    obs_vr = header['OBS_VR']
    obs_vw = header['OBS_VW']
    obs_vn = header['OBS_VN']
    fgh_tsp = header['FGH_TSP1']
    contposn = header['CONTPOSN']

    with open(output_file, 'a') as f:
        f.write(f'{date}, {did}, {dsun_au}, {car_rot}, {crln_obs}, {hgln_obs}, {crlt_obs}, {obs_vr}, {obs_vw}, {obs_vn}, {fgh_tsp}, {contposn}\n')
```
