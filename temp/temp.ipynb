```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
import glob
```

```python
files = sorted(glob.glob('/home/ulyanov/data/4science/hrt/*'))
files[0], files[-1], len(files)
```

```python
plt.ioff()

for file in files[:]:
    print(file)
    figfile = file.split('/')[-1]

    with fits.open(file) as hdul:
        header = hdul[0].header.copy()
        data = hdul[0].data.copy()

    fig, ax = plt.subplots(figsize=(8,8))
    ax.imshow(data, 'seismic', vmin=-200, vmax=200, origin='lower')
    ax.set_title(figfile)

    fig.tight_layout()
    fig.savefig(f'pics/{figfile}.png')
    plt.close(fig)


plt.ion()
```
