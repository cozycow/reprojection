```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from datetime import datetime
import glob
from reprojection import *
from show import *
```

```python
files = np.array(sorted(glob.glob('/home/ulyanov/data/solo/phi/2026/blos_/*')))

dates = np.array([datetime.fromisoformat(file.split('/')[-1].split('_')[-3]) for file in files])

t = np.where(np.all([dates > datetime(2026,7,22),
                     dates < datetime(2026,9,29)], axis=0))[0] # CR2314 + CR2315

files = files[t]
files[0], files[-1], len(files)
```

```python
for rsun_deg in np.arange(-12, -89, -1):
    data, view = make_map(files[:], correct_dr=True, kind='warp', mu_thr=0.1, rsun_deg=rsun_deg, powmu=0, crln=0)
    show_data(data, view, label=r'$B_{r}$, G', cmap='hmimag', vmin=-200, vmax=200, to_folder='png')
```
