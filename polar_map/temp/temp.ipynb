```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
```

```python
def crn(date, crln):
    dt = date - datetime(1853, 11, 9, 0, 0, 0)

    cdn = dt.total_seconds() / 24 / 3600
    crn = cdn / 27.2753 + 1

    print(crn)
    delta = crn + crln / 360
    delta -= np.round(delta)

    return  crn - delta

```

```python
crln = np.arange(0,360)
date = datetime(2026,7,23)
```

```python
plt.figure(figsize=(12,8))
plt.plot(crln, crn(date, crln))
```
